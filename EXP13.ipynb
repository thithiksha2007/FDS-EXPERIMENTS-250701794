```python
import matplotlib.pyplot as cricket

Overs = [5, 10, 15, 20, 25, 30, 35, 40, 45, 50]
India = [30, 55, 90, 129, 165, 200, 239, 270, 310, 350]
SriLanka = [25, 70, 90, 120, 140, 170, 195, 220, 255, 279]

cricket.plot(Overs, India, color="green", marker="o", label="INDIA")
cricket.plot(Overs, SriLanka, color="red", marker="s", label="SRILANKA")

cricket.title("INDIA Vs SRILANKA")
cricket.xlabel("Overs")
cricket.ylabel("Score")
cricket.legend(loc="upper left")
cricket.grid(True)


```

```python
import matplotlib.pyplot as hscmark
import numpy as np

Students = ['SHREE', 'DEV', 'KEERTHI', 'PRIYA', 'SHAN', 'KUMARAN']
Marks = [96, 91, 94, 75, 45, 81]

x = np.arange(len(Students))

hscmark.bar(x, Marks, color=['green', 'blue', 'orange', 'pink', 'red', 'purple'])

hscmark.xticks(x, Students, rotation=45)
hscmark.xlabel("Student Names")
hscmark.ylabel("HSC Percentage")
hscmark.title("HSC Mark Comparison", fontsize=18, color="blue")

hscmark.ylim(0, 100)
hscmark.grid(axis="y", linestyle="--", alpha=0.5)

hscmark.show()
```

```python
import matplotlib.pyplot as election

Candidates = ['CANDIDATE 1', 'CANDIDATE 2', 'CANDIDATE 3', 'CANDIDATE 4']
Votes = [315, 130, 245, 210]

Colors = ['blue', 'gold', 'green', 'pink']
Explode = (0, 0.1, 0, 0)

election.pie(
    Votes,
    labels=Candidates,
    colors=Colors,
    explode=Explode,
    autopct='%1.1f%%',
    startangle=90,
    shadow=True
)

election.title("Election Vote Percentage")
election.axis('equal')
election.show()
```
